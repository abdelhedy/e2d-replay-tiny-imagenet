## Cell 5 — Run Training (First Time)

**Before running:** update `TINY_DATA_ROOT` to the path found in Cell 3.  
For example: `/kaggle/input/tiny-imagenet` or `/kaggle/input/tiny-imagenet-200`

```python
import subprocess, os, shutil, torch

os.chdir('/kaggle/working')

if os.path.exists('./e2d_cl_output'):
    shutil.rmtree('./e2d_cl_output')
    print('Deleted old e2d_cl_output — starting fresh.')

TINY_DATA_ROOT = '/kaggle/input/datasets/akash2sharma/tiny-imagenet'

TEACHERS = (
    '/kaggle/input/datasets/abdelhedifarah/e2d-tiny-code/tinyimagenet/rn18_200ep.pth,'
    '/kaggle/input/datasets/abdelhedifarah/e2d-tiny-code/tinyimagenet/rn50_200ep.pth'
)

STUDENT_TINY_CKPT = (
    '/kaggle/input/datasets/abdelhedifarah/e2d-tiny-code/tinyimagenet/rn18_200ep.pth'
)

gpu_ids = ','.join(str(i) for i in range(torch.cuda.device_count())) or '0'
print(f'Using GPU(s): {gpu_ids}')

result = subprocess.run([
    'python', 'cl_train.py',

    '--dataset',             'tiny',
    '--tiny-data-root',      TINY_DATA_ROOT,
    '--tiny-normalization',  'imagenet',

    '--strategy',            'e2d',
    '--n-experiences',       '5',
    '--n-classes',           '20',
    '--seed',                '42',

    '--student-arch',        'resnet18',
    '--student-tiny-checkpoint', STUDENT_TINY_CKPT,
    '--freeze-backbone',

    '--epochs',              '50',
    '--lr',                  '0.05',
    '--batch-size',          '256',

    '--ipc',                 '10',
    '--fixed-per-class',

    '--K',                   '150',
    '--recover-iterations',  '300',
    '--recover-lr',          '0.01',
    '--recover-batch-size',  '16',

    '--relabel-views',       '5',
    '--relabel-temperature', '20.0',
    '--relabel-batch-size',  '256',

    '--teacher-checkpoints', TEACHERS,

    '--kd-loss',             'dist',
    '--kd-weight',           '1.0',
    '--kd-temperature',      '4.0',

    '--gpu-id',              gpu_ids,
    '--output-dir',          './e2d_cl_output',

    '--debug-diagnostics',
], text=True)

print('Return code:', result.returncode)
```

```python
## Cell — Download Output as ZIP
import zipfile, os
from pathlib import Path
from IPython.display import FileLink, display

src = Path('/kaggle/working/e2d_cl_output')
out = '/kaggle/working/e2d_cl_output.zip'

if not src.exists():
    print('ERROR: e2d_cl_output not found.')
else:
    with zipfile.ZipFile(out, 'w', zipfile.ZIP_DEFLATED) as zf:
        for f in src.rglob('*'):
            if f.is_file():
                zf.write(f, f.relative_to(src.parent))
    size_mb = os.path.getsize(out) / 1e6
    print(f'Zipped {src} → {out}  ({size_mb:.1f} MB)')
    display(FileLink(out, result_html_prefix='Download: '))
```

```python
from pathlib import Path

root = Path('/kaggle/input/datasets/akash2sharma/tiny-imagenet/tiny-imagenet-200')
wnids = [x.strip() for x in (root / 'wnids.txt').read_text().splitlines()]
folders = sorted([p.name for p in (root / 'train').iterdir() if p.is_dir()])

print('First 20 wnids.txt classes:')
print(wnids[:20])
print('\nFirst 20 ImageFolder sorted classes:')
print(folders[:20])
print('\nSame first 20?', wnids[:20] == folders[:20])

mismatch = [(i, w, f) for i, (w, f) in enumerate(zip(wnids, folders)) if w != f]
print('First mismatches:', mismatch[:10])
```
